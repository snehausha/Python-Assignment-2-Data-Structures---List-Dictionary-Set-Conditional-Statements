### 2. List Operations / Modifications:

 a. Appended the string "Yazhini" to name_list using '**.append()**'

 [Adds an element at the end of the list.]

 b. Inserted the element 30 at index 2 in age_list using '**.insert()**'

 [places the element at the specified position.]

 c. Removed the string "Yazhini" from name_list using '**.remove()**'

 [Deletes the specified element.]

 d. Popped out the last element from age_list using '**.pop()**'

 [Remove the last element from the list]

 e. Extended the age_list with additional ages [29, 30, 26] using '**.extend()**'

 [Adds multiple elements]

 f. Sorted age_list in descending order using '**.sort(reverse=True)**'

 [Arranges elements in descending order]

 g. Finded Max age, Min age and sum of all ages from age_list using '**max(),min(),sum()**'

 [max()- prints the largest element

min()-prints the smallest element

```python
name_list.append("Yazhini")
print(name_list)

age_list.insert(2,30)
print(age_list)


name_list.remove("Yazhini")
print(name_list)

age_list.pop()
print(age_list)

age_list.extend([29,30,26])
print(age_list)

age_list.sort(reverse=True)
print(age_list)

print("Maximum age:",max(age_list))
print("Minimum age:",min(age_list))
print("Sum of ages:",sum(age_list))

```

a. Created a dictionary named student_marks

[.keys() → returns all keys (student names).

## TASK 4:Operators & Conditional Statements :(IF, ELIF, ELSE)

Performance Category Program:

1. Prompt user for Input Score range: **'INPUT()'**
2. performance category criteria:

a. Above Average: Score greater than 7

b. Average: Score between 4 and 7(both inclusive)

c. Below Average: Score lesser than 4


3. Output: Print the Performance category: '**print()**'

4. Given  prompt to each category:

Above Average:(Excellent work,Keep it up!)

Average:(Good effort!Keep practicing, there's room for improvement.)

Below Average:(Performance need improvement,Practice consistently)



```python
score=float(input("Enter your score(0-10): "))


if score>7 and score<=10:
  print("Performance:ABOVE AVERAGE")
  print("Remarks: Excellent work,Keep it up!")
elif score>4 and score<=7:
  print("Performance:AVERAGE")
  print("Remarks: Good effort!Keep practicing, there's room for improvement.")
elif score>=0 and score<=4:
  print("Performance:BELOW AVERAGE")
  print("Remarks: Performance need improvement,Practice consistently")
else:
   print("INVALID SCORE")



```
